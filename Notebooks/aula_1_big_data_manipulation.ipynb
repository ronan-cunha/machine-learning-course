## 1. Entendendo os Metadados do Arquivo


```python
import os
import pandas as pd
```

```python
file_path = "/content/machine_learning_course/deputies_dataset.csv"

# A. Tamanho do arquivo no disco
file_size_bytes = os.path.getsize(file_path)
file_size_gb = file_size_bytes / (1024 ** 3)

print(f"Tamanho do arquivo: {file_size_gb:.2f} GB")
```

```python

# B. Estimativa rápida do número total de linhas (sem carregar o arquivo na RAM)
# Lendo o arquivo linha por linha via buffer de texto
with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
    total_lines = sum(1 for line in f)

print(f"Total estimado de linhas: {total_lines:,}")
```

```python
# Lendo apenas a linha 0 (cabeçalho) ajustando nrows=0
df_header = pd.read_csv(file_path, nrows=0)
colunas = df_header.columns.tolist()

print(f"Total de colunas: {len(colunas)}")
print("Nomes das colunas:")
for i, col in enumerate(colunas):
    print(f" [{i}] {col}")
```

```python
# Carrega as 100 primeiras linhas com todas as colunas
df_sample = pd.read_csv(file_path, nrows=100)

print(f"Formato da amostra: {df_sample.shape}")
print("\nVisualização das primeiras 5 linhas:")
display(df_sample.head(10))
```

```python
print("\nTipos de dados:")
print(df_sample.dtypes)
```

```python
# Suponha que desejamos analisar apenas a coluna "deputy_id"
coluna_alvo = ["deputy_id"]

df_coluna = pd.read_csv(file_path, usecols=coluna_alvo)

print("Estrutura da coluna importada:")
print(df_coluna.info())
print("\nPrimeiros valores:")
print(df_coluna.head())
```
